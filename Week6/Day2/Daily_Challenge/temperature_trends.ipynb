# Daily Challenge: NumPy, Pandas, and Matplotlib Integration

**Scenario**: You are a data analyst working with a dataset of global weather. Your task is to analyze temperature trends and visualize the results.

## 1. Data Preparation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

np.random.seed(42)

# Generate synthetic monthly temperature data for 10 cities over 12 months
temperature_data = np.random.uniform(low=-5, high=35, size=(10, 12))

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
cities = [f'City {i}' for i in range(1, 11)]

df = pd.DataFrame(temperature_data, index=cities, columns=months)

print('Dataset shape:', df.shape)
df
```

## 2. Data Analysis

```python
# Annual average temperature for each city
annual_avg = df.mean(axis=1)

print('Annual Average Temperature (°C) per City:')
print(annual_avg.round(2).to_string())

hottest_city = annual_avg.idxmax()
coldest_city = annual_avg.idxmin()

print(f'\nCity with the HIGHEST average temperature: {hottest_city} ({annual_avg[hottest_city]:.2f} °C)')
print(f'City with the LOWEST  average temperature: {coldest_city} ({annual_avg[coldest_city]:.2f} °C)')
```

## 3. Data Visualization

```python
fig, ax = plt.subplots(figsize=(14, 7))

# Use a green-toned color palette
greens = plt.cm.BuGn(np.linspace(0.35, 0.95, len(cities)))

for idx, city in enumerate(cities):
    linestyle = '--' if city in [hottest_city, coldest_city] else '-'
    linewidth = 2.5 if city in [hottest_city, coldest_city] else 1.2
    ax.plot(months, df.loc[city], color=greens[idx],
            linestyle=linestyle, linewidth=linewidth, label=city)

ax.set_title('Monthly Temperature Trends for Each City', fontsize=16, pad=15)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Temperature (°C)', fontsize=12)
ax.axhline(0, color='grey', linewidth=0.8, linestyle=':')
ax.yaxis.grid(True, linestyle='--', alpha=0.6)
ax.set_axisbelow(True)
ax.legend(loc='lower left', fontsize=9, framealpha=0.8)

plt.tight_layout()
plt.savefig('temperature_trends.png', dpi=150)
plt.show()
print('Chart saved as temperature_trends.png')
```

## 4. Summary Report

### Findings

- **Dataset**: Synthetic monthly temperatures (°C) for **10 cities** over **12 months**, uniformly distributed between **-5 °C and 35 °C**.
- The city with the **highest** annual average temperature will be highlighted above after running the analysis cell.
- The city with the **lowest** annual average temperature is similarly identified.

### Interesting Trends

- Because the data is randomly generated (uniform distribution), no true seasonal pattern exists; each city fluctuates independently across months.
- The dashed lines in the chart highlight the two extreme cities (hottest and coldest), making them easy to distinguish from the rest.
- In a real-world dataset we would expect clear seasonal sinusoidal patterns — hotter summers and colder winters for cities in temperate zones.

### Methodology

| Step | Tool used |
|---|---|
| Data generation | `np.random.uniform` |
| DataFrame creation | `pd.DataFrame` |
| Annual average | `DataFrame.mean(axis=1)` |
| Extreme cities | `idxmax()` / `idxmin()` |
| Visualization | `matplotlib.pyplot` |
